```python
import numpy as np
import math
import matplotlib.pyplot as plt
from decimal import *
```

```python
c = 299792458 # m/s
def relativistic_mom(energy,mass):
    # E^2 = p^2 + m^2
    # m is in eV/c
    # E is in eV
    p = math.sqrt(energy**2-mass**2)
    return p
```

```python
delm12 = 0.759*10**(-4) # eV^2/c^2
delm13 = 23.2*10**(-4) # eV^2/c^2
```

```python
mass1 = 0.8 # 0.8 or below in eV/c
```

```python
# p = \gamma m v
# \gamma = 1/\sqrt(1-v^2/c^2)
# v/c = 1/\sqrt(1-1/\gamma^2)
#  v^2 = p^2/m^2 * (1-v^2/c^2)
# (1 + p^2/m^2 1/c^2) v^2 = p^2/m^2
# v = p m c / \sqrt(c^2 + p^2 / m^2)
v1 = relativistic_mom(10**9,mass1)/mass1
v2 = relativistic_mom(10**9,(mass1+math.sqrt(delm12)))/(mass1+math.sqrt(delm12))
v3 = relativistic_mom(10**9,(mass1+math.sqrt(delm13)))/(mass1+math.sqrt(delm13))
v1 = v1*c/math.sqrt(c**2+v1**2)
v2 = v2*c/math.sqrt(c**2+v2**2)
v3 = v3*c/math.sqrt(c**2+v3**2)
```

```python
getcontext().prec = 40
```

```python
v1=((Decimal(1)-Decimal(mass1**2) / Decimal(10**18))*Decimal(c))
v2=((Decimal(1)-Decimal((mass1+math.sqrt(delm12))**2) / Decimal(10**18))*Decimal(c))
v3=((Decimal(1)-Decimal((mass1+math.sqrt(delm13))**2) / Decimal(10**18))*Decimal(c))
print(v1,v2,v3)
print(Decimal(mass1**2) / Decimal(10**18))
print(Decimal((mass1+math.sqrt(delm12))**2) / Decimal(10**18))
print((Decimal(1)-Decimal((mass1+math.sqrt(delm12))**2) / Decimal(10**18))*Decimal(c))
```

```python
plt.plot(np.arange(10000)*10**(-9),float(v1)*np.arange(10000)*10**(-9))
plt.title("distance traveled (m) as function of time (s)")
```

```python
plt.plot(np.arange(1000)*10**(-9),float(v1-v2)*np.arange(1000)*10**(-9))
plt.plot(np.arange(1000)*10**(-9),float(v1-v3)*np.arange(1000)*10**(-9))
plt.title('separation with some hierarchy')
```
